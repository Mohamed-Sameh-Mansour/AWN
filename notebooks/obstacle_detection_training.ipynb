```python
import os
import shutil
import random
import yaml
from collections import Counter

# =================================================================
# 1. القائمة الـخارقة والنهائية للمشروع (12 كلاس منظمين تماماً)
# =================================================================
FINAL_CLASSES = {
    0: "Crosswalk",      # ممرات المشاة
    1: "Edge_Sidewalk",  # الرصيف وحواف الطرق والمشايات
    2: "Opened_Door",    # أبواب السيارات المفتوحة
    3: "Traffic_Light",  # إشارات المرور وعلامات التوقف
    4: "Obstacle",       # العوائق (صناديق القمامة، الكراسي، الحواجز، الأشجار)
    5: "Vehicle",        # المركبات (سيارات، حافلات، شاحنات، دراجات، قطارات)
    6: "Person",         # البشر (أشخاص، وجوه)
    7: "Pole_Bollard",   # الأعمدة الرأسية والأعمدة القصيرة على الرصيف
    8: "Pothole",        # الحفر والتشققات الأرضية
    9: "Stairs",         # السلالم
    10: "Animal",        # الحيوانات (كلاب، قطط، حيوانات عامة)
    11: "Cane"           # العصا البيضاء للمكفوفين
}

# =================================================================
# 2. خرائط التحويل الجراحية (تطابق الـ IDs الفعلية 100% في ملفاتك)
# =================================================================
map_d1 = {0:0, 1:1, 2:2, 3:3, 11:3, 4:4, 5:5, 12:5, 6:6, 7:7, 8:8, 10:9}
map_d2 = {1:0, 9:1, 11:3, 2:4, 8:4, 13:5, 15:5, 4:6, 5:7, 6:8, 10:9, 0:10}
map_d3 = {0:4, 4:4, 12:4, 1:5, 3:5, 13:5, 2:7, 8:7, 5:2, 6:6, 7:1, 10:1, 9:8, 11:9}
map_d4 = {1:0, 6:1, 8:3, 2:4, 5:4, 9:4, 10:5, 3:6, 4:7, 0:10, 7:9}

# الداتا الخامسة (تم الاكتفاء بكلاسات الشارع والحيوانات والعصا وحذف كلاسات المطبخ والمنزل تلقائياً)
map_d5 = {
    24:3, 29:3,       # stop sign, traffic lights -> Traffic_Light
    1:4, 3:4, 8:4, 31:4, # barrier, bench, chair, trash bin -> Obstacle
    4:5, 5:5, 6:5, 17:5, 30:5, 32:5, # bicycle, bus, car, motorcycle, train, truck -> Vehicle
    20:8,             # pothole -> Pothole
    23:9,             # stairs -> Stairs
    7:10, 9:10,       # cat, dog -> Animal
    11:6, 12:6,       # human face recognition, human obstacle detection -> Person
    15:11             # mobility cane -> Cane
}

# عداد إجمالي للكشف عن توزيع البيانات الجديد ومنع الـ Overfitting
final_distribution = Counter()

def merge_and_filter_dataset(src_base, dest_base, mapping, dataset_name, split="train"):
    src_img_dir = os.path.join(src_base, split, "images")
    src_lbl_dir = os.path.join(src_base, split, "labels")
    dest_img_dir = os.path.join(dest_base, split, "images")
    dest_lbl_dir = os.path.join(dest_base, split, "labels")

    os.makedirs(dest_img_dir, exist_ok=True)
    os.makedirs(dest_lbl_dir, exist_ok=True)

    if not os.path.exists(src_lbl_dir):
        return 0

    copied_images = 0
    img_extensions = ['.jpg', '.jpeg', '.png', '.JPG', '.PNG']

    for label_file in os.listdir(src_lbl_dir):
        if not label_file.endswith('.txt'):
            continue

        with open(os.path.join(src_lbl_dir, label_file), 'r') as f:
            lines = f.readlines()

        new_lines = []
        has_rare_class = False
        has_heavy_class = False
        temp_counts = []

        for line in lines:
            parts = line.split()
            if not parts: continue
            old_id = int(parts[0])

            # إذا كان الكلاس الحالي مدعوم في خطتنا (وليس من أدوات المطبخ/المنزل المستبعدة)
            if old_id in mapping:
                new_id = mapping[old_id]
                parts[0] = str(new_id)
                new_lines.append(" ".join(parts) + "\n")
                temp_counts.append(new_id)

                # تصنيف من أجل الموازنة ومنع الـ Overfitting
                if new_id in [5, 6]:  # سيارات أو أشخاص
                    has_heavy_class = True
                if new_id in [0, 2, 8, 9, 10, 11]: # الكلاسات النادرة والخطيرة (حفر، حيوانات، سلالم، عصا، أبواب)
                    has_rare_class = True

        if new_lines:
            # 💡 تطبيق منطق الموازنة الصارم:
            # لو الصورة في الترين تحتوي فقط على سيارات/أشخاص بدون أي عوائق أو حفر، نحتفظ بـ 15% منها عشوائياً ونحذف الباقي
            if split == "train" and has_heavy_class and not has_rare_class:
                if random.random() > 0.15:
                    continue # تجاهل الصورة

            # البحث عن الصورة المقابلة لملف التكست لنسخها
            base_name = os.path.splitext(label_file)[0]
            img_file = None
            for ext in img_extensions:
                if os.path.exists(os.path.join(src_img_dir, base_name + ext)):
                    img_file = base_name + ext
                    break

            if img_file:
                # حفظ ملف التكست الجديد المعدل بالـ IDs الموحدة
                with open(os.path.join(dest_lbl_dir, label_file), 'w') as f:
                    f.writelines(new_lines)
                # نسخ الصورة المقابلة
                shutil.copy(os.path.join(src_img_dir, img_file), os.path.join(dest_img_dir, img_file))
                copied_images += 1

                # إضافة التكرارات للعداد العام لفحص النتيجة النهائية (في الـ train فقط)
                if split == "train":
                    for n_id in temp_counts:
                        final_distribution[FINAL_CLASSES[n_id]] += 1

    print(f"✅ {dataset_name} ({split}): تم دمج وتصفية {copied_images} صورة.")
    return copied_images

# المجلد الموحد والنهائي للتدريب
OUTPUT_DATASET = "/content/Blind_Navigation_Dataset"

all_datasets = [
    {"path": "dataset_1_road_house", "map": map_d1, "name": "Road House (1)"},
    {"path": "dataset_2_safewalkbd", "map": map_d2, "name": "SafeWalkBD (2)"},
    {"path": "dataset_3_smart_glasses", "map": map_d3, "name": "Smart Glasses (3)"},
    {"path": "dataset_4_outdoor", "map": map_d4, "name": "Outdoor Det (4)"},
    {"path": "dataset_5_object_det", "map": map_d5, "name": "Object Det (5)"}
]

print("⏳ جاري بدء الصهر الجراحي للخمس داتا سيتس وتطبيق قواعد الموازنة والاتزان...")
print("-" * 70)

for d in all_datasets:
    if os.path.exists(d["path"]):
        merge_and_filter_dataset(d["path"], OUTPUT_DATASET, d["map"], d["name"], split="train")
        merge_and_filter_dataset(d["path"], OUTPUT_DATASET, d["map"], d["name"], split="valid") # مجلد الـ val
    else:
        print(f"❌ تنبيه: المجلد {d['path']} غير موجود، يرجى التأكد من تحميله أولاً.")

# كتابة ملف الإعدادات الموحد data.yaml للـ YOLOv11
yaml_content = {
    'train': f"{OUTPUT_DATASET}/train/images",
    'val': f"{OUTPUT_DATASET}/valid/images",
    'nc': len(FINAL_CLASSES),
    'names': list(FINAL_CLASSES.values())
}
with open(f"{OUTPUT_DATASET}/data.yaml", 'w') as f:
    yaml.dump(yaml_content, f, default_flow_style=False)

# =================================================================
# 📊 طباعة تقرير الاتزان النهائي الموحد بعد التصفية
# =================================================================
print("\n" + "="*50)
print("📊 التوزيع الإجمالي والنهائي للكلاسات في داتا الـ Train المدمجة:")
print("="*50)
for idx in sorted(FINAL_CLASSES.keys()):
    name = FINAL_CLASSES[idx]
    count = final_distribution.get(name, 0)
    print(f" 🔹 كلاس {idx:<2} | {name:<15} ➡️ مكرر {count:<6} مرة")
print("="*50)
print("🎉 مبروك! الداتا سيت فائقة القوة 'Blind_Navigation_Dataset' جاهزة الآن 100% للتدريب.")
```

```python
import os
import cv2
import random
import yaml
import matplotlib.pyplot as plt
import numpy as np

# 1. مسارات الداتا سيت المدمجة الجديدة
DATASET_DIR = "/content/Blind_Navigation_Dataset"
images_dir = os.path.join(DATASET_DIR, "train", "images")
labels_dir = os.path.join(DATASET_DIR, "train", "labels")

# القائمة الموحدة للكلاسات وأسمائها
FINAL_CLASSES = {
    0: "Crosswalk", 1: "Edge_Sidewalk", 2: "Opened_Door", 3: "Traffic_Light",
    4: "Obstacle", 5: "Vehicle", 6: "Person", 7: "Pole_Bollard", 8: "Pothole",
    9: "Stairs", 10: "Animal", 11: "Cane"
}

# دالة ذكية ومعدلة لتحويل إحداثيات YOLO (صناديق أو مضلعات) ورسمها بأمان
def draw_yolo_box(img, box_data, class_name):
    h, w, _ = img.shape
    class_id = int(box_data[0])
    coords = box_data[1:]

    # 💡 إذا كانت الإحداثيات عبارة عن صندوق تقليدي (4 قيم بعد الـ ID)
    if len(coords) == 4:
        x_center, y_center, bbox_w, bbox_h = coords
        x1 = int((x_center - bbox_w / 2) * w)
        y1 = int((y_center - bbox_h / 2) * h)
        x2 = int((x_center + bbox_w / 2) * w)
        y2 = int((y_center + bbox_h / 2) * h)

    # 💡 إذا كانت الإحداثيات مضلع Segmentation (أكثر من 4 قيم)
    elif len(coords) > 4:
        # فصل نقاط X ونقاط Y (النقاط الزوجية هي X والفردية هي Y)
        xs = coords[0::2]
        ys = coords[1::2]
        # حساب أطراف الصندوق المحيط بالمضلع (Bounding Box Extraction)
        x1 = int(min(xs) * w)
        x2 = int(max(xs) * w)
        y1 = int(min(ys) * h)
        y2 = int(max(ys) * h)
    else:
        return img # إذا كانت البيانات غير صالحة

    # رسم المستطيل والنص بأمان
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
    cv2.putText(img, class_name, (x1, max(y1 - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)
    return img

# تجهيز الشكل العام للعرض (شبكة 4 صفوف × 3 أعمدة لتستوعب 12 كلاس)
fig, axes = plt.subplots(4, 3, figsize=(15, 18))
axes = axes.ravel()

print("🔍 جاري فحص البيانات مرئياً بالاعتماد على محرك القراءة المرن الجديد...")

# امتدادات الصور المدعومة
img_extensions = ['.jpg', '.jpeg', '.png', '.JPG', '.PNG']

for class_id, class_name in FINAL_CLASSES.items():
    found_image = False
    all_label_files = os.listdir(labels_dir)
    random.shuffle(all_label_files) # خلط عشوائي للملفات للبحث

    for label_file in all_label_files:
        if not label_file.endswith('.txt'): continue

        with open(os.path.join(labels_dir, label_file), 'r') as f:
            lines = f.readlines()

        target_box = None
        for line in lines:
            parts = line.split()
            if parts and int(parts[0]) == class_id:
                target_box = [float(x) for x in parts]
                break

        if target_box:
            base_name = os.path.splitext(label_file)[0]
            img_path = None
            for ext in img_extensions:
                test_path = os.path.join(images_dir, base_name + ext)
                if os.path.exists(test_path):
                    img_path = test_path
                    break

            if img_path:
                img = cv2.imread(img_path)
                if img is not None:
                    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    img = draw_yolo_box(img, target_box, class_name)

                    axes[class_id].imshow(img)
                    axes[class_id].set_title(f"Class {class_id}: {class_name}", fontsize=12, color='darkgreen')
                    axes[class_id].axis('off')
                    found_image = True
                    break

    if not found_image:
        axes[class_id].text(0.5, 0.5, f"No image found for\n{class_name}",
                            ha='center', va='center', fontsize=12, color='red')
        axes[class_id].axis('off')

plt.tight_layout()
plt.show()
```

```python
import os
import torch
import yaml

print("🕵️‍♂️ بدء الفحص الهندسي الشامل قبل التدريب...")
print("-" * 50)

# 1. فحص الـ GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"💻 عتاد التدريب: تم كشف جهاز من نوع ⬅️ ({device.upper()})")
if device == "cuda":
    print(f"   🔹 اسم الكارت: {torch.cuda.get_device_name(0)}")
    torch.cuda.empty_cache() # تنظيف الكاش القديم فوراً

# 2. فحص ملف data.yaml
yaml_path = "/content/Blind_Navigation_Dataset/data.yaml"
if os.path.exists(yaml_path):
    with open(yaml_path, 'r') as f:
        config = yaml.safe_load(f)
    nc = config.get('nc', 0)
    names = config.get('names', [])
    print(f"📄 فحص الـ YAML: عدد الكلاسات المسجلة ({nc}) | عدد الأسماء الفعلية ({len(names)})")
    if nc != len(names):
        print("❌ خطأ: هناك عدم تطابق بين عدد الكلاسات وأسمائها!")
    else:
        print("✅ ملف الـ YAML سليم ومطابق تماماً.")
else:
    print("❌ خطأ: ملف data.yaml غير موجود في المسار المحدد!")

# 3. فحص الإحداثيات والملفات التالفة (عيّنة من الترين)
labels_dir = "/content/Blind_Navigation_Dataset/train/labels"
out_of_bounds_count = 0
corrupted_lines = 0

if os.path.exists(labels_dir):
    for file in os.listdir(labels_dir)[:2000]: # نفحص أول 2000 ملف كعينة إحصائية سريعة
        if file.endswith('.txt'):
            with open(os.path.join(labels_dir, file), 'r') as f:
                for line in f.readlines():
                    parts = line.split()
                    if not parts: continue
                    try:
                        # جلب الإحداثيات بعد الـ ID
                        coords = [float(x) for x in parts[1:]]
                        # التحقق أن القيم بين 0 و 1 (فقط للصناديق القياسية أو النقاط الموحدة)
                        if any(c < 0.0 or c > 1.0 for c in coords):
                            out_of_bounds_count += 1
                    except ValueError:
                        corrupted_lines += 1

    print(f"🎯 فحص البيانات: ")
    print(f"   🔹 الأسطر التالفة برمجياً: {corrupted_lines}")
    print(f"   🔹 بوكسات خارج الحدود (Out of Bounds): {out_of_bounds_count}")
    if out_of_bounds_count == 0 and corrupted_lines == 0:
        print("✅ الإحداثيات سليمة بنسبة 100% وضمن الحدود المنطقية [0, 1].")
    else:
        print("⚠️ تنبيه: توجد بعض القيم الطفيفة خارج الحدود، سيقوم YOLO بمعالجتها تلقائياً، لا تقلق.")

print("-" * 50)
print("🚀 النتيجة النهائية: البيئة والداتا جاهزون هندسياً للانطلاق!")
```
